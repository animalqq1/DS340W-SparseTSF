```python
from pathlib import Path

path = Path("exp/exp_main.py")
text = path.read_text()

old1 = """        preds = []
        trues = []
        inputx = []
        folder_path = './test_results/' + setting + '/'
        if not os.path.exists(folder_path):
            os.makedirs(folder_path)
"""

new1 = """        # Calculate metrics without storing the entire test set.
        sum_abs_error = 0.0
        sum_sq_error = 0.0
        sum_true = 0.0
        sum_true_sq = 0.0
        num_values = 0
"""

old2 = """                outputs = outputs.detach().cpu().numpy()
                batch_y = batch_y.detach().cpu().numpy()

                pred = outputs  # outputs.detach().cpu().numpy()  # .squeeze()
                true = batch_y  # batch_y.detach().cpu().numpy()  # .squeeze()

                preds.append(pred)
                trues.append(true)
                inputx.append(batch_x.detach().cpu().numpy())
                if i % 20 == 0:
                    input = batch_x.detach().cpu().numpy()
                    gt = np.concatenate((input[0, :, -1], true[0, :, -1]), axis=0)
                    pd = np.concatenate((input[0, :, -1], pred[0, :, -1]), axis=0)
                    visual(gt, pd, os.path.join(folder_path, str(i) + '.pdf'))
"""

new2 = """                pred = outputs.detach().cpu()
                true = batch_y.detach().cpu()
                diff = pred - true

                sum_abs_error += diff.abs().sum(dtype=torch.float64).item()
                sum_sq_error += diff.square().sum(dtype=torch.float64).item()
                sum_true += true.sum(dtype=torch.float64).item()
                sum_true_sq += true.square().sum(dtype=torch.float64).item()
                num_values += true.numel()
"""

old3 = """        preds = np.array(preds)
        trues = np.array(trues)
        inputx = np.array(inputx)

        preds = preds.reshape(-1, preds.shape[-2], preds.shape[-1])
        trues = trues.reshape(-1, trues.shape[-2], trues.shape[-1])
        inputx = inputx.reshape(-1, inputx.shape[-2], inputx.shape[-1])

        # result save
        folder_path = './results/' + setting + '/'
        if not os.path.exists(folder_path):
            os.makedirs(folder_path)

        mae, mse, rmse, mape, mspe, rse, corr = metric(preds, trues)
"""

new3 = """        # result save
        folder_path = './results/' + setting + '/'
        if not os.path.exists(folder_path):
            os.makedirs(folder_path)

        mae = sum_abs_error / num_values
        mse = sum_sq_error / num_values
        true_sse = sum_true_sq - (sum_true * sum_true) / num_values
        rse = np.sqrt(sum_sq_error) / np.sqrt(max(true_sse, 1e-12))
"""

for old, new in [(old1, new1), (old2, new2), (old3, new3)]:
    if old not in text:
        raise RuntimeError("Expected code block was not found.")
    text = text.replace(old, new, 1)

path.write_text(text)
print("Memory-safe testing installed successfully.")
```

```python
from pathlib import Path

path = Path("/content/DS340W-SparseTSF/exp/exp_main.py")
text = path.read_text()

start = text.index("    def test(self, setting, test=0):")
end = text.index("    def predict(self, setting, load=False):", start)

backup = path.with_name("exp_main_before_memory_patch.py")
if not backup.exists():
    backup.write_text(text)

new_test_method = '''    def test(self, setting, test=0):
        test_data, test_loader = self._get_data(flag='test')

        if test:
            print('loading model')
            checkpoint_path = os.path.join(
                './checkpoints/' + setting,
                'checkpoint.pth'
            )
            self.model.load_state_dict(
                torch.load(checkpoint_path, map_location=self.device)
            )

        folder_path = './test_results/' + setting + '/'
        os.makedirs(folder_path, exist_ok=True)

        sum_abs_error = 0.0
        sum_sq_error = 0.0
        sum_true = 0.0
        sum_true_sq = 0.0
        num_values = 0

        self.model.eval()

        with torch.no_grad():
            for i, (batch_x, batch_y, batch_x_mark, batch_y_mark) in enumerate(test_loader):
                batch_x = batch_x.float().to(self.device)
                batch_y = batch_y.float().to(self.device)

                outputs = self.model(batch_x)

                f_dim = -1 if self.args.features == 'MS' else 0
                outputs = outputs[:, -self.args.pred_len:, f_dim:]
                batch_y = batch_y[:, -self.args.pred_len:, f_dim:]

                diff = (outputs - batch_y).double()
                true_values = batch_y.double()

                sum_abs_error += diff.abs().sum().item()
                sum_sq_error += diff.square().sum().item()
                sum_true += true_values.sum().item()
                sum_true_sq += true_values.square().sum().item()
                num_values += true_values.numel()

                if i % 20 == 0:
                    input_values = batch_x[0, :, -1].detach().cpu().numpy()
                    true_plot = batch_y[0, :, -1].detach().cpu().numpy()
                    pred_plot = outputs[0, :, -1].detach().cpu().numpy()

                    gt = np.concatenate((input_values, true_plot), axis=0)
                    pd = np.concatenate((input_values, pred_plot), axis=0)
                    visual(gt, pd, os.path.join(folder_path, str(i) + '.pdf'))

                if self.args.test_flop:
                    test_params_flop((batch_x.shape[1], batch_x.shape[2]))
                    return

        mae = sum_abs_error / num_values
        mse = sum_sq_error / num_values
        rmse = np.sqrt(mse)

        true_sse = sum_true_sq - (sum_true * sum_true) / num_values
        rse = np.sqrt(sum_sq_error) / np.sqrt(max(true_sse, 1e-12))

        print('mse:{}, mae:{}, rse:{}'.format(mse, mae, rse))

        results_path = './results/' + setting + '/'
        os.makedirs(results_path, exist_ok=True)

        np.save(
            os.path.join(results_path, 'metrics.npy'),
            np.array([mae, mse, rmse, rse])
        )

        with open("result.txt", 'a') as f:
            f.write(setting + " \\n")
            f.write('mse:{}, mae:{}, rse:{}'.format(mse, mae, rse))
            f.write('\\n\\n')

        return

'''

path.write_text(text[:start] + new_test_method + text[end:])
print("Memory-safe testing method installed.")
```
